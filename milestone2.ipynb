```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# This is the data used for the visualization
data_incident_plot = pd.read_csv("Incidents_Year.csv")

# again, just inspecting the data
data_incident_plot.head()
```

```python
# creating the canvas size
plt.figure(figsize=(20, 10)) 

# setting the x variable
x = range(len(data_incident_plot["Incidents"]))

# plotting
plt.plot(x, data_incident_plot["Incidents"]) 
plt.xticks(x, data_incident_plot["Year"]) 
plt.xlabel("Year") 
plt.ylabel("Extralegal Collective Incidents Resulting in Fatalities") 
plt.xticks(rotation = 90) 
plt.title("figure 1: Incidents of Extralegal Collective Violence Resulting in Fatalities, 1783-1865")

plt.show()
```

```python
# second plot
data_killed_plot = pd.read_csv("Killed_Year.csv")

data_killed_plot.head()
```

```python
plt.figure(figsize=(20, 10)) 

# setting the x variable
x = range(len(data_killed_plot["Killed"]))

# plotting
plt.plot(x, data_killed_plot["Killed"]) 
plt.xticks(x, data_killed_plot["Year"]) 
plt.xlabel("Year") 
plt.ylabel("Recorded Number of Fatalities in Extralegal Collective Action")
plt.xticks(rotation = 90) 
plt.title("figure 2: Recorded Number of Fatalities in Extralegal Collective Action, 1783-1865")

plt.show()
```

The journal was transparent and provided a GitHub repository where all the data for the visualizations could be found. The data that was used for these specific visualizations was organized well so that the csv files only had the year and the number of killed as well as the number of recorded incidents. For the purpose of data visualizations, having these simple csv’s made it easier for me to plot.  

For both plots, I used the matplot  library to create the visualizations. Again, since the csv’s were simple, I knew that I had to assign the year to the x axis and the number of incidents or killings to the y axis. I also created the variable x that counted the number of rows that needed to be plotted to make sure the number of each row value was represented on the plot.  

While strictly looking at the visualizations, I found the difference between “Incidents of Extralegal Collective Violence Resulting in Fatalities” and “Recorded Number of Fatalities in Extralegal Collective Action” difficult to understand. To understand this difference, the author of the journal explains that figure one measures the number of instances that resulted in at least one killing, but it does not tell us how many people were killed exactly. The second graph tells us exactly how many people were killed in the events of extralegal killing. In other words, figure one tells us the number of violent events and how frequent the events were, whereas figure two tells us the number of people killed and how deadly it was. 

Having these two differences is important because it can show us as historians that even if there were more events, this does not mean that they were more deadly, and on the other hand it can also shows us that less events does not mean that less people were killed. Having these distinctions is important when analyzing this data, so we don’t make false conclusions, and the author of this journal makes this clear through these visualizations.
